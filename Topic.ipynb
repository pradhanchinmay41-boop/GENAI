#### How GenAi works :
*  It works with two main key features :  GAN & VAE .

    *GAN (Generative adverserial neural network)
    * gan are class of ML model .

     * it used widely for generative task such as image synthesis ,style transfor .

    * The key idea behind GANs is to have two neural network (a)- a generator , (b) - a discriminator .

    * (a) Generator -- it take random data and try to create real data .

       * generator pull the data from internet or take from users .

    * (b) discriminator -- it is a separete neural network that evaluates wheather  a given input is real(from the training data ) or fake(generated by generator) .
      * it acctually bunnary classification .

##### workflow of GANs : Generator --Discriminator -- Adversarial Training -- Training proces 

##### VAE(variational auto encoder) :
* It has Two architecture :

                       * 1. Encoder architecture -- input

                        * It maps the input data into lower dimmensional representation that helps captuering essesntial thing about this topic . 
                        * It give from user side to the prompt .

                      * 2. decoder architecture -- output 

                        * Transformer respond to the prompt with low error . 


* Auto encoder  - it is a type of artificial neural network used for unsupervised learning .
