### **Model Finetunning**


* **Fine tuning** enables you to take a pre trained model and adapt it to perform better for a specific usecse by training it on your own data .

* It is a powerful technique that improves model performances on specific tasks .

* Fine-tunning works by training an existing pre-trained models .

* Fine-tune Techniques :

                    * 1 : Full parameter finetunning(FPFT )

                          * apply all of the model's parameters . The full model is loaded into memory andthe weights are adjusted byshowing the model . 

                    * 2 : Parameter efficient fine-tunning(PEPT)

                          * It is the Technique that allows us to fine tune only a subset of the model's parameters, both reducing the amount of compute required for each .

                    * 3 : Low-rank Adaption(LoRA)

                          * It is a **PEPT** technique that reduces the no.of trainable parameter by adding low-rank matrices to the network and only training those. LoRA results in a significant reduction in the number oftrainable parameters making it more efficient in terms of both compute and memory requirements .

                          * Here we reduce memory . 

                          * Advantages :

                                  * Reduce memory requirements(Require less memory to store gradients andoptimizer states)

                                  * Faster training(reduce the computational cost of fine tunning)

                                  * Improved performance(It can achieve comparable or even better performnce than full fine tunning)

                          * Disadvantages :

                          * It may struggle to fully adapt themodel to tasks that require large changes in internal representations .



                    * 4 : Quantized low rank adaptaion(QLoRA)

                          * It is an extension ofLoRA that further reduces the memory requirements by quantizing the model's weights and activations . it
                           use a combination of quantization and low rank adaption to achieve even more efficient fine-tuning .

                          * Here we reduces memory and weights .

                          * Advantages : 
                                  * Even lower memory requirements(reduce the memory requirements even further by quantizing the model's weight and activations ,making possible to fine=tune large modelin limited hardware .)
                                            
                                  * Increased efficiency(It acieve a better tradeoff between memory requirement and performances )

                                
                    * Reinforcement learning from human feedback(RLHF)
                          

                    * Reinforcement learning from verifiable rewards(RLVR)

                   

* Advantages of finetunning :

                   * Reduce token size

                   * Domain specific knowlewdge

                   * Improved performance 

* **Finetune Technique** means by reduce model weights and quantized to 1b ,4b, 8bit which can be run in local machine also .

* quantization = reducing precision by representing values with fewer bits or fewer discrete levels.

* A model's weights might be stored as 32-bit floating-point numbers (FP32)Quantization can convert them to 8-bit integers (INT8) or 4-bit integers (INT4).

####  **Quantization technique** 

* ## 🔹 1. bitsandbytes (bnb)

👉 Most widely used for **LLM quantization**

**What it does:**

* Converts model weights into **8-bit or 4-bit**
* Reduces GPU memory drastically

**Key Features:**

* ✅ 8-bit (LLM.int8)
* ✅ 4-bit (NF4 – used in QLoRA)
* ✅ Works with Hugging Face Transformers
* ✅ Supports training + inference

**Example:**

```python
from transformers import BitsAndBytesConfig

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4"
)
```

**Use Case:**
👉 QLoRA, low-memory training, single GPU setups

## 🔹 2. Accelerate

👉 Not a quantization method — it's a **training/inference orchestration tool**

**What it does:**

* Handles:

  * Multi-GPU
  * Mixed precision (fp16/bf16)
  * Device placement
* Works with quantization libraries like bitsandbytes

**Key Features:**

* ✅ Easy scaling across GPUs
* ✅ Handles memory efficiently
* ❌ Does NOT quantize itself

**Use Case:**
👉 Combine with bitsandbytes / DeepSpeed / FSDP

---

## 🔹 3. GPTQ (Post-Training Quantization)

👉 High-performance inference quantization

**What it does:**

* Quantizes model **after training**
* Mostly **4-bit**

**Key Features:**

* ✅ Very fast inference
* ✅ Minimal accuracy loss
* ❌ Not used for training

**Libraries:**

* AutoGPTQ

**Use Case:**
👉 Production inference (chatbots, APIs)

---

## 🔹 4. AWQ (Activation-aware Weight Quantization)

👉 Newer and more accurate than GPTQ

**What it does:**

* Uses activations to preserve important weights

**Key Features:**

* ✅ Better accuracy than GPTQ
* ✅ Optimized for GPUs
* ❌ Mostly inference-focused

**Use Case:**
👉 High-quality inference with low memory

---

## 🔹 5. QLoRA (Technique, not library)

👉 Combines quantization + LoRA

**What it does:**

* Uses:

  * bitsandbytes (4-bit)
  * LoRA adapters
* Fine-tunes large models on small GPUs

**Key Features:**

* ✅ Train 7B–65B models on single GPU
* ✅ Memory efficient
* 🔥 Most important modern technique

---

## 🔹 6. FP16 / BF16 (Mixed Precision)

👉 Not quantization, but related optimization

**What it does:**

* Uses 16-bit instead of 32-bit

**Key Features:**

* ✅ Faster training
* ✅ Less memory
* ❌ Not as efficient as 4-bit

##### **2. Tempreture**

* Range of tempreture: 0 - 2 .

* Tempreture is a parameter that controls the randomness of the model's output during text generation.

* using helpof temp. LLM modelpull simple/understanable word from internet . 

* It decide how model are creative .

* **low Tempreture(eg-0.2)** - This model generates more predictable and focused response . resulting in coherent and sensible output but less creativity .

* **Coherent** means logical, clear, well-organized, and easy to understand.

* **Medium Tempreture(eg-0.7)** - This setting strikes a balance between randomness and predictibility . allowing for some creativity while  still producing reasonable coherent responses .

* **HIgh Tempreture(eg-1.0)** - This model produce more diverse and unpredictable output .

* Best Tempreture is bydefault 1(means medium) .

* **One line ans** - Tempreture generate coherent response from internet so that the end user easily understand the output .  


##### **3. Top-p**

* It is a parameter that control how the model select the nextword during text generation .

* Top-p and tempreture are both parameter used to control the randomness and creativity of text generation .

* Temp and Top-p both are bestfriend .


* Low temperature: Predictable, coherent responses.

* High temperature: More creative and varied, but can lead to less coherent outputs.

* Lower p: More focused responses, as only the highest probability words are included.

*  Higher p: Allows for a wider range of words to be selected, increasing creativity.

* Low Temperature Low Top-p: Very focused and coherent responses, suitable for tasks needing precision.

* Low Temperature High Top-p: Still focused but with some creative variation.

* High Temperature + Low Top-p: Creative but can risk coherence, as the model might pick from a smaller set of choices.

* High Temperature High Top-p: Maximizes creativity and diversity, but at the potential cost of coherence.

* **Medium temp + Max top-p is best .**

##### **6. Frequency penalty**
