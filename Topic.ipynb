# **LLM model :**

* It is text based model .
* How does LLM work :

                  * 1. LLM Tokenizer

                  * 2. LLM Embedding

                  * 3. Transfer learning

                  * 4. Attention in LLM 

                  * 5. Pre-trainning 

* Nlp and LLm tokenizer both are different concept . 

* Nlp tokenizer create  word  as token but LLM split the sentence to text or subword which are converted ti ID's represent in entry in embedding matrix  that is used model for represent for specific token .  

* LLM tokenizer divided into 3 parts : 

                                 * 1. BPE(byte pair encoding)

                                 * 2. Word piece

                                 * 3. Sentence piece

*  1. BPE(byte pair encoding) :

   * in this what ever u content provide it divided into indivisual character , thn check it will be 2 time or 3 time  . it will continue till reach the right requirement vocabulary .
   * Manually it can't possible so we introduce a library called hugging face . 
   * In BPE we choose frequency .

* 2. Word piece :

  * very simmilar to BPE . in this we will choose likehood for the training data added to vocabulary instead of frequency.
  * In BPE we got frequency but in wordpiece we will be using probability frequency . 

*  3. Sentence piece :

   *  It create space between two word ,coz if there is no space between word it can't be consider as token.
   * it used in TS /ALBERT/ROBERT etc . 


* 2. LLM Embedding : 
   * Embedding are vectors stored in an index within a vector database . 
   * Embedding are a way to store data of all types in number arrays called vector .

* Whatever u have content  it convert to embadding and store in vector db ,than what ever user ask this query also store in vectordb thn apply LLM model thn llm model retrive correct response  it is called llm embedding .
